# VENV CREATION/ACTIVATION

```python
import venv, os, subprocess, sys

VENV_DIR = os.path.join(os.getcwd(), "bank_mkt_venv")
#venv.create(VENV_DIR, with_pip=True)

PYTHON = os.path.join(VENV_DIR, "Scripts", "python.exe")
print(f"Venv Python: {PYTHON}")
print(f"Kernel Python: {sys.executable}")  # Will differ — that's expected
```

# FOLDER STRUCTURE

```python
import os
import nbformat

dirs = [
    "app",
    "notebooks"
]
# 
notebooks = [
   "notebooks/01_data_prep.ipynb", #3
    "notebooks/02_model_training.ipynb", #3
    "notebooks/03_model_serving.ipynb", #3
    "notebooks/04_drift_monitoring.ipynb", #3
    "app/main.py", #3
    "requirements.txt",
    ".gitattributes",
    "README.md"
]

for folder in dirs:
    os.makedirs(folder, exist_ok=True)

for nb_path in notebooks:
    nb = nbformat.v4.new_notebook()
    with open(nb_path, "w") as f:
        nbformat.write(nb, f)
        
```

# INSTALL DEPENDENCIES

```python
import subprocess

result = subprocess.run(
    [PYTHON, "-m", "pip", "install", "-r", "requirements.txt"],
    capture_output=True,
    text=True
)

print(result.stdout)
```

```python
# SANITY CHECK: List installed packages in the venv to confirm installation
result = subprocess.run(
    [PYTHON, "-m", "pip", "list"],
    capture_output=True, text=True
)
print(result.stdout)
```
